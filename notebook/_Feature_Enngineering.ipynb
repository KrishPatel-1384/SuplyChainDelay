# 04. Feature Engineering & Pipelines
## Objectives:
- Engineer order duration & transit metrics.
- One-Hot/Target Encoding for categorical variables (`shipping_mode`, `market`, `customer_segment`).
- Build standard Scikit-Learn Preprocessing Pipeline.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('/home/patel-krish/anaconda3/BrainyBeam/SuplyChainDelay/data/syndelay_v1.csv').drop(columns=['label'])
```

```python
df
```

```python
# Select categorical & numerical features
```

```python
from sklearn.compose import ColumnTransformer
```

```python
num_cols = ['sales', 'profit_per_order', 'order_item_quantity', 'order_item_discount_rate', 
            'product_price', 'sales_per_customer', 'latitude', 'longitude']
```

```python
cat_cols = ['shipping_mode', 'market', 'customer_segment', 'department_name']
```

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
    ]
)
```

```python
X_train_proc = preprocessor.fit_transform(df[num_cols + cat_cols])
print("Processed Feature Array Shape:", X_train_proc.shape)
```

```python
# Saving the processor pipepline 
```

```python
os.makedirs('../models', exist_ok=True)
joblib.dump(preprocessor, '../models/preprocessor.joblib')
print("Preprocessor pipeline saved successfully to models/preprocessor.joblib")
```
