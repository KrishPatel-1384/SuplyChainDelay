# 05. Model Training (XGBoost & Random Forest)
## Objectives:
- Train multi-class XGBoost Classifier.
- Train Random Forest Classifier for baseline comparison.
- Save trained artifacts into the models/directory.

```python
import pandas as pd
import joblib
```

```python
df = pd.read_csv('/home/patel-krish/anaconda3/BrainyBeam/SuplyChainDelay/data/syndelay_v1.csv')
```

```python
num_cols = ['sales', 'profit_per_order', 'order_item_quantity', 'order_item_discount_rate', 
            'product_price', 'sales_per_customer', 'latitude', 'longitude']
cat_cols = ['shipping_mode', 'market', 'customer_segment', 'department_name']
```

```python
X = df[num_cols + cat_cols]
y = df['label']
```

```python
preprocessor = joblib.load('/home/patel-krish/anaconda3/BrainyBeam/SuplyChainDelay/models/preprocessor.joblib')
X_proc = preprocessor.transform(X)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_proc, y, test_size=0.2, random_state=42, stratify=y)
```

```python
# XGBoost Model
```

```python
from xgboost import XGBClassifier
```

```python
xgb_model = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    random_state=42,
    eval_metric='mlogloss'
)
xgb_model.fit(X_train, y_train)
joblib.dump(xgb_model, '/home/patel-krish/anaconda3/BrainyBeam/SuplyChainDelay/models/xgboost_delay_model.joblib')
```

```python
# Random Forest Model
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(n_estimators=100, max_depth=12, random_state=42)
rf_model.fit(X_train, y_train)
joblib.dump(rf_model, '/home/patel-krish/anaconda3/BrainyBeam/SuplyChainDelay/models/random_forest_delay_model.joblib')

print("Models trained and saved successfully!")
```

```python
# Save LightGBM model artifact
```

```python
from lightgbm import LGBMClassifier
```

```python
lgb_model = LGBMClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
)
lgb_model.fit(X_train, y_train)

joblib.dump(lgb_model, '/home/patel-krish/anaconda3/BrainyBeam/SuplyChainDelay/models/lightgbm_delay_model.joblib')
print("LightGBM model trained and saved successfully to models/lightgbm_delay_model.joblib")
```
